```python
import pandas as pd
```

```python
df = pd.read_csv('employees.csv')
```

```python
df.head()
```

```python

df_joined = df.merge(
    df, 
    left_on='Referred By ID', 
    right_on='Employee ID', 
    suffixes=('_caller', '_other'),
    how='left'
)

df['Name'] = df_joined['Name_caller']
df['Referred By Name'] = df_joined['Name_other']
```

```python
df.head()
```

```python
df.drop('Referred By ID', axis=1, inplace=True)
```

```python
df.to_csv('employees.csv', index=False)
```
